```python
def amortization_schedule(principal, annual_rate, years, extra=0.0):
    i, n = annual_rate / 12, years * 12
    pmt = principal * i / (1 - (1 + i) ** -n)
    rows, bal = [], principal
    for k in range(1, n + 1):
        interest = bal * i
        amort = min(pmt - interest + extra, bal)
        bal -= amort
        rows.append((k, interest + amort, interest, amort, bal))
        if bal <= 1e-8:
            break
    return pd.DataFrame(rows, columns=["mes", "cuota", "interes", "amortizacion", "saldo"]).set_index("mes")

base = amortization_schedule(principal, annual_rate, years)
base.head()
```

## Ejercicio 2 — Pagos extra
¿Cuántos meses e intereses se ahorran con 200 €/mes adicionales?

```python
fast = amortization_schedule(principal, annual_rate, years, extra=200)
print(f"Meses: {len(base)} → {len(fast)}")
print(f"Intereses: {base.interes.sum():,.0f} → {fast.interes.sum():,.0f} (ahorro {base.interes.sum()-fast.interes.sum():,.0f})")
pd.DataFrame({"base": base.saldo, "extra": fast.saldo}).plot(title="Saldo pendiente"); plt.show()
```
