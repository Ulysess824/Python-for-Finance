## 1. La cartera

```python
rng = np.random.default_rng(3)
N = 10_000
P = rng.lognormal(np.log(15), 0.6, N)                 # monto, millones de pesos
rate = rng.uniform(0.12, 0.30, N)                     # tasa efectiva anual simple /12
n = rng.choice([12, 24, 36, 48, 60, 72], N)           # plazo en meses
loans = pd.DataFrame({"monto": P, "tasa": rate, "plazo": n})
display(loans.describe().round(2))
print(f"Cartera total: {P.sum():,.0f} millones")
```

Cuota francesa: $\;PMT = P\,\dfrac{i}{1-(1+i)^{-n}}\;$ con $i$ = tasa mensual. Saldo tras $k$ pagos: $\;B_k = P(1+i)^k - PMT\,\dfrac{(1+i)^k-1}{i}$.
